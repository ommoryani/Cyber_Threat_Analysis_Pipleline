#### Unzipping the file and storing it in a folder 

```python
import zipfile
zip_file_path="archive.zip"
output_folder="extracted_files"
try:
    with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
        zip_ref.extractall(output_folder)
    print(f"✅ Complete: File extracted successfully into the {output_folder}!")
except zipfile.BadZipFile:
    print("❌ Error: The file is not a valid ZIP archive.")
except FileNotFoundError as e:
    print(f"❌ An unexpected error occurred: {e}")
```

#### Understanding the data

```python
import pandas as pd
file_path = "extracted_files/cybersecurity synthesized data.csv"
try:
    df = pd.read_csv(file_path)
    print("Data Loaded Into Python Successfully!\n")
    print("\n")
    print("📋 [SHAPE CHECK]")
    print(f"Total Security Log Entries (Rows): {df.shape[0]:,}")
    print(f"Total Metrics Measured (Columns): {df.shape[1]}")
    print("-" * 50)
    print("👥 [DUPLICATE CHECK]")
    print(f"Total Duplicate Rows Found: {df.duplicated().sum()}")
    print("-" * 50)
    print("📂 [DATA TYPE SCAN]")
    print(df.info())
    print("-" * 50)
    print("🔍 [MISSING DATA SCAN]")
    missing_data = df.isnull().sum()
    if missing_data.sum() == 0:
        print("Perfect! Zero missing values detected across all columns.")
    else:
        print(missing_data[missing_data > 0])

except FileNotFoundError:
    print(f"⚠️ Error: Could not locate the CSV file at '{file_path}'.")
    print("💡 Double-check the spelling inside your 'data' folder.")
```

#### Data Optimization

```python
if 'timestamp' in df.columns:
    df['timestamp'] = pd.to_datetime(df['timestamp'], errors='coerce')
    df['Incident Date'] = df['timestamp'].dt.date
    df['Incident Year'] = df['timestamp'].dt.year
    df['Incident Month'] = df['timestamp'].dt.strftime('%B')
    df['Incident Hour'] = df['timestamp'].dt.hour
    df = df.drop(columns=['timestamp'])
    print("⏰ Step 1 Complete: Timestamps cleanly split into Date, Year, Month, and Hour!")
else:
    print("ℹ️ 'timestamp' column not found; skipping timestamp conversion.")
df.info()
df.head()
print("⏰ Optimization 1 Complete: Timestamps cleanly split into Date, Year, Month, and Hour!")
```

```python
rename_mapping = {
    'attack_type': 'Attack Type',
    'target_system': 'Target System',
    'outcome': 'Outcome',
    'attacker_ip': 'Attacker IP',
    'target_ip': 'Target IP',
    'data_compromised_GB': 'Data Compromised (GB)',
    'attack_duration_min': 'Attack Duration (Min)',
    'security_tools_used': 'Security Tool Used',
    'user_role': 'User Role',
    'location': 'Country',
    'attack_severity': 'Attack Severity Scale',
    'industry': 'Industry',
    'response_time_min': 'SOC Response Time (Min)',
    'mitigation_method': 'Mitigation Method'
}
df = df.rename(columns=rename_mapping)
print("📋 Optimization 2 Complete: Column headers optimized for corporate reporting!")
df.info()
```

```python
ext_cols = df.select_dtypes(include=['str', 'object']).columns
for col in ext_cols:
    df[col] = df[col].astype(str).str.strip()
print("🧼 Step 3 Complete: Text entries cleanly trimmed of whitespace!")
df.head(10)
```

```python

text_columns = df.select_dtypes(include=['str', 'object']).columns
print("🔍 --- CASE SENSITIVITY ANOMALY SCAN --- 🔍\n")
any_issues_found = False
for col in text_columns:
    if 'Date' in col or 'IP' in col or 'Month' in col:
        continue
        
    raw_unique_values = df[col].dropna().unique()
    seen_lower = {}
    duplicates = []
    
    for val in raw_unique_values:
        val_str = str(val).strip()
        val_lower = val_str.lower()
        
        if val_lower in seen_lower:
            if val_str not in seen_lower[val_lower]:
                seen_lower[val_lower].append(val_str)
                if val_lower not in duplicates:
                    duplicates.append(val_lower)
        else:
            seen_lower[val_lower] = [val_str]
    if duplicates:
        any_issues_found = True
        print(f"⚠️ Column '{col}' contains case format inconsistencies!")
        print(f"   Conflicting Text Structures Found: ")
        for d in duplicates:
            print(f"   -> {seen_lower[d]}")
            for variant in seen_lower[d]:
                row_count = df[df[col] == variant].shape[0]
                print(f"      * '{variant}' appears in {row_count} rows")
        print("-" * 60)

if not any_issues_found:
    print("✅ Absolute Clarity! Every single categorical text column is completely uniform in casing across all 100,000 rows.")
    print("   There are no hidden duplicate strings hiding behind mixed capitalizations!")
```

```python

categorical_columns = ['Attack Type', 'Target System', 'Outcome', 'Security Tool Used', 'User Role', 'Industry', 'Mitigation Method']

print("🔍 --- DEEP CATEGORICAL TEXT SCAN --- 🔍\n")

for col in categorical_columns:
    unique_vals = df[col].unique()
    print(col)
    print(f"Number of unique categories: {len(unique_vals)}")
    print(f"Values: {list(unique_vals)}")
    print("-" * 60)
```

```python

text_cols = df.select_dtypes(include=['str', 'object']).columns

for col in text_cols:
    if 'IP' in col:
        df[col] = df[col].astype(str).str.strip().str.lower()
    else:
        df[col] = df[col].astype(str).str.strip().str.title()

print(" Step 4 Complete: All text columns standardized to uniform Title Case!")

# 3. Verify the Country entries look pristine now
print("\n🌍 Cleaned unique entries in Country column:")
print(df['Country'].unique())
```
